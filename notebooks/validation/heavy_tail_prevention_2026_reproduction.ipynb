Compare independent estimates with the authors' CSVs using `python examples/compare_heavy_tail_paper.py output/prevention_paper_reproductions.json PRIVATE_FIGURES_DIRECTORY`; combined MC errors, not identical random seeds, are the relevant diagnostic. Annual strict ruin is not compound-Poisson ruin. Zero light losses allow exact event skipping; nonzero light losses retain annual updates. The simulator fails explicitly if any path is censored; a truncated mean is never presented as E[tau]. Theoretical prevention ratios compare proportional changes, not equal intervention costs. Finite-mean comparisons use equal removed annual mean loss and a common cost, and simulate the exact all-time PK supremum rather than a finite-horizon proxy.
